## Step 1: Set Up Environment Variables & Client
Securely load your OpenAI API key and instantiate the OpenAI client.

```python
import os
from getpass import getpass
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

if "OPENAI_API_KEY" not in os.environ or not os.environ["OPENAI_API_KEY"]:
    os.environ["OPENAI_API_KEY"] = getpass("Enter your OpenAI API Key: ")

client = OpenAI()
EMBED_MODEL = "text-embedding-3-small"
CHAT_MODEL = "gpt-4o-mini"
```

## Step 2: The Knowledge Cutoff
With no tools, no web search, and no system date injected, a plain model call can only draw on what it learned during training. It has no clock and no live connection to the outside world, so it cannot know today's date.

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": "what is today's date."}],
)

print(response.choices[0].message.content)
```

---
## Step 3: Classification & Named Entity Recognition (NER)
A chat model can perform classic NLP tasks directly from a prompt — no training required. Here we use `gpt-4o-mini` for:
* **Intent Classification** — sorting a user message into a fixed set of categories.
* **Named Entity Recognition** — extracting structured entities (people, orgs, locations, dates) as JSON.

```python
def run_llm_task(task_name, prompt, system_prompt="You are a helpful NLP assistant."):
    print(f"\n--- Running Task: {task_name} ---")
    response = client.chat.completions.create(
        model=CHAT_MODEL,
        messages=[
            {'role': 'system', 'content': system_prompt},
            {'role': 'user', 'content': prompt}
        ]
    )
    print(response.choices[0].message.content)
```

```python
# Intent Classification
user_msg = "I need to reset my password, I forgot it."
run_llm_task("Intent Classification", f"Classify this into [Support, Sales, Billing]: {user_msg}")
```

```python
# Named Entity Recognition (NER)
ner_text = "Apple CEO Tim Cook visited Mumbai on April 18, 2026."
run_llm_task("NER", f"Extract (Person, Org, Location, Date) as JSON from: {ner_text}")
```
