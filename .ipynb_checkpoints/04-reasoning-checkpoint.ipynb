# Reasoning vs. Non-Reasoning Models (OpenAI API)

This notebook demonstrates the behavioral and API differences between:
1. **Non-reasoning model (`gpt-4o-mini`)**: Fast, direct inference without background Chain-of-Thought (CoT).
2. **Reasoning model (`o3-mini`)**: Uses background reasoning tokens controlled by `reasoning_effort` to explore logic paths before answering.

```python
!pip install --upgrade openai
```

```python
import os
from openai import OpenAI

# Initialize OpenAI client (requires OPENAI_API_KEY environment variable)
client = OpenAI()

# Logic puzzle prompt
prompt = """
A room has 3 light switches outside, connected to 3 light bulbs inside the room.
You cannot see inside the room from outside.
You are allowed to flip any switches, but you can only enter the room ONCE.

How can you figure out which switch controls which bulb?
Answer step-by-step succinctly.
"""

print("=" * 70)
print(f"PROMPT:\n{prompt.strip()}")
print("=" * 70)
```

## 1. Non-Reasoning Model (`gpt-4o-mini`)

```python
print("[1] Calling Non-Reasoning Model (gpt-4o-mini)...\n")

response_non_reasoning = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": prompt}
    ]
)

print("--- GPT-4o-MINI RESPONSE ---")
print(response_non_reasoning.choices[0].message.content)
print(f"\nTotal tokens used: {response_non_reasoning.usage.total_tokens}")
```

## 2. Reasoning Model (`o3-mini`)

```python
print("[2] Calling Reasoning Model (o3-mini with High Reasoning Effort)...\n")

# o3-mini and o1 support reasoning_effort: "low", "medium", or "high"
response_reasoning = client.chat.completions.create(
    model="o3-mini",
    reasoning_effort="high",
    messages=[
        {"role": "user", "content": prompt}
    ]
)

print("--- O3-MINI RESPONSE ---")
print(response_reasoning.choices[0].message.content)

# Inspect token metrics
usage = response_reasoning.usage
print("\n--- TOKEN METRICS ---")
print(f"Total tokens: {usage.total_tokens}")
if hasattr(usage, "completion_tokens_details") and usage.completion_tokens_details:
    print(f"Reasoning tokens spent in background: {usage.completion_tokens_details.reasoning_tokens}")
```
