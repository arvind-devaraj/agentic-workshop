# Reasoning vs. Non-Reasoning Models (OpenAI API)

This notebook demonstrates the behavioral and API differences between:
1. **Non-reasoning model (`gpt-4o-mini`)**: Fast, direct inference without background Chain-of-Thought (CoT).
2. **Reasoning model (`o3-mini`)**: Uses background reasoning tokens controlled by `reasoning_effort` to explore logic paths before answering.

```python
import os
from getpass import getpass
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

if "OPENAI_API_KEY" not in os.environ or not os.environ["OPENAI_API_KEY"]:
    os.environ["OPENAI_API_KEY"] = getpass("Enter your OpenAI API Key: ")

client = OpenAI()
EMBED_MODEL = "text-embedding-3-small"
CHAT_MODEL = "gpt-4o-mini"
```

```python
import os
from openai import OpenAI

client = OpenAI()

time_prompt = """  How many positive integers less than 25 are divisible by neither 3 nor 5? Final number only.
"""

print("PROMPT:")
print(time_prompt.strip())
print("=" * 70)

# 1. Non-Reasoning Model (gpt-4o-mini)
print("\n[1] NON-REASONING MODEL (gpt-4o-mini):\n")
res_non_reasoning = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": time_prompt}]
)
print(res_non_reasoning.choices[0].message.content)

# 2. Reasoning Model (o3-mini)
print("\n" + "=" * 70 + "\n")
print("[2] REASONING MODEL (o3-mini):\n")
res_reasoning = client.chat.completions.create(
    model="o3-mini",
    reasoning_effort="high",
    messages=[{"role": "user", "content": time_prompt}]
)
print(res_reasoning.choices[0].message.content)
```
